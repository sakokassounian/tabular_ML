## Recap on precision, recall, and f1-score, ROC-AUC

### 🎯 Precision

* **What it measures**: Of all the predicted positive cases, how many are actually positive.
* **Use case**: Important when **false positives** are costly (e.g., spam detection, important emails are blocked).
* **Formula**:

  $$
  \text{Precision} = \frac{\text{True Positives (TP)}}{\text{True Positives (TP)} + \text{False Positives (FP)}}
  $$

---



### 🔍 Recall 

* **What it measures**: Of all actual positive cases, how many were correctly predicted.
* **Use case**: Important when **false negatives** are costly (e.g., disease detection, the person can die).
* **Formula**:

  $$
  \text{Recall} = \frac{\text{True Positives (TP)}}{\text{True Positives (TP)} + \text{False Negatives (FN)}}
  $$

---


### ⚖️ F1-Score

* **What it measures**: The **harmonic mean** of precision and recall — it balances both metrics.
* **Use case**: Best when you need a **trade-off between precision and recall**, especially with **imbalanced datasets**.
* **Formula**:

  $$
  \text{F1-score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}
  $$

* A **high F1-score** means both **precision** and **recall** are reasonably high.
* It ranges from **0** (worst) to **1** (perfect).
* ⚠️ Warning:
    * Ignores true negatives, making it less suitable for imbalanced datasets.
    * Assumes equal importance of precision and recall, which may not fit all problems.
    * Not ideal when different types of errors (false positives vs false negatives) have different real-world costs.
  

---

### 🤔 Decision process

<img src="./data/metrics.png" alt="VAE Structure" width="600"/>



* **Recall-focused model:** Catches almost every cancer case (few false negatives) but mislabels more healthy people as sick (more false positives), exposing them to unnecessary, risky treatments.
  
* **Precision-focused model:** Gives highly reliable cancer diagnoses (few false positives), yet misses some true cases (more false negatives), which can be fatal.
  
* Because missing a cancer patient is worse than an unnecessary treatment, practitioners usually set the decision threshold to favor **high recall** over precision.
* Before training, always decide which error type—false positives or false negatives—carries the greater real-world cost and tune the model to minimize that cost.
* The tradeoff between precision and recall shifts with the classification threshold, and while F1-score peaks when both are balanced, accuracy remains mostly unchanged—highlighting its inadequacy as a reliable metric in graph example given above.

### 🔔 Receiver Operating Characteristic (ROC) curve

The ROC curve, originally developed in WWII to improve radar signal detection, is now widely used in machine learning to evaluate classifiers. It plots the True Positive Rate against the False Positive Rate at various classification thresholds, helping to visualize the tradeoff between correctly identifying positives and avoiding false alarms. The ROC AUC quantifies this performance, guiding threshold selection for optimal model accuracy.

* True Positive rates (TPR) = TP / P = TP / (TP+FP)   [True alaram rate - Precision] <br>* TPR -> maximized (TP -> max, FP -> min)

* False positive rate (FPR) = FP / N = FP / (FP+TN)   [False alarm rate - Type I error] <br>* FPR -> minimzed  (FP -> min, TN -> max)  


<img src="./data/ROC-AUC.png" alt="VAE Structure" width="600"/>

A good ROC curve stays close to the top-left corner, showing high true positives with few false positives — this means the model is performing well. A diagonal line means the model is guessing randomly and is no better than chance. The ROC curve also helps compare models and choose the best threshold.



When comparing models with ROC curves, anything above the diagonal (random guess) is decent, and the closer to the top-left, the better. A result that's too perfect or exactly random likely signals issues—like data leakage or a poorly framed problem. Real-world models usually fall in between

* **AUC** tells you how well your model can **rank** a positive example **above** a negative one.
* If you pick one positive and one negative at random, **AUC is the chance** the model gives the higher score to the positive.

  * **AUC = 1.0** → always ranks correctly ✅
  * **AUC = 0.5** → ranks randomly 🎲
  * **AUC = 0.0** → always ranks incorrectly ❌

This is **similar to the Wilcoxon-Mann-Whitney test** in statistics.

---

### ⚠️ Things to Keep in Mind

* AUC **doesn’t tell you the actual probabilities** — just the ranking.
* It **treats all mistakes equally**, which may not be good in critical cases like **cancer detection**, where **false negatives** are more serious.
* Still, plotting the **ROC curve** helps visualize how your model behaves across thresholds.

---

In short: **AUC is great for ranking, but not always ideal when the cost of errors varies**.
