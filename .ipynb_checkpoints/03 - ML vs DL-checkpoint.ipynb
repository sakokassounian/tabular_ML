# Chapter 3: Machine Learning vs. Deep Learning for Tabular Data

## Intorudction

In this chapter, we explore two competing approaches to solving tabular data problems: **classical machine learning** (e.g., XGBoost) and **deep learning** (e.g., Keras). We'll compare them across three key dimensions:

* **Simplicity** – How easy it is to implement and understand
* **Transparency** – How explainable the models are
* **Efficacy** – How well they perform in real-world tasks

Using two examples—predicting Airbnb listing prices in NYC and estimating property selling times—we’ll analyze where each approach shines and where it falls short. This chapter aims to help you choose the right tool for your tabular data problems.

---

To predict whether a New York City Airbnb listing is priced above or below the average, we use a tabular dataset containing \~49,000 real listings from NYC, available on Kaggle. This dataset is ideal for comparing machine learning and deep learning because:

* It's large enough to be meaningful but small enough for easy handling.
* It contains a balanced number of columns—suitable for both ML and DL.
* It's easily viewable and explorable in tools like spreadsheets.
* It features diverse column types: continuous, categorical, geospatial, and text.
* It has minor imperfections (like missing values) but is clean enough to model without heavy preprocessing.
* It’s open-source and based on real business data.

The target variable is derived from the `price` column, classifying listings as above or below the median price.

## Summary and conclusions



- We compared machine learning (ML) and deep learning (DL) using three things:
  - **Simplicity** – how easy the code is to write and understand
  - **Transparency** – how easy it is to explain the model and see what features matter
  - **Efficacy** – how well the model performs<br></br>

- **ML is simpler** to code and maintain.
- **ML is more transparent**, making it easier to explain to others.
- **Efficacy is similar** for both ML and DL out of the box, but tuning could make a difference.
- It’s still unclear which one is better for things like:
  - Winning Kaggle competitions
  - Business use
  - Research popularity
